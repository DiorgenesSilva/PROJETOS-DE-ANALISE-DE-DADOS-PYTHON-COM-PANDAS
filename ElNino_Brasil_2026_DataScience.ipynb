## 📖 Glossário de Siglas

| Sigla | Significado | O que é |

| **INPE** | Instituto Nacional de Pesquisas Espaciais | Instituto de pesquisa brasileiro (Ministério da Ciência e Tecnologia) |
| **CPTEC** | Centro de Previsão de Tempo e Estudos Climáticos | Braço do INPE responsável pela modelagem numérica e previsões climáticas de médio/longo prazo |

| **ONI** | Oceanic Niño Index | Índice oficial (média trimestral móvel da anomalia de TSM na região Niño 3.4) usado para classificar El Niño/La Niña |
| **TSM** | Temperatura da Superfície do Mar | Variável monitorada para detectar anomalias associadas ao ENSO |
| **ENSO** | El Niño–Southern Oscillation | Ciclo climático que alterna entre El Niño (aquecimento) e La Niña (resfriamento) no Pacífico Equatorial |
| **SMN** | Servicio Meteorológico Nacional (Argentina) | Agência oficial argentina de meteorologia e clima |

## 🌡️ Seção 2 — Base de Dados: Episódios de El Niño Muito Forte / Super El Niño (1972–2026)
> **Leigo:** Uma tabela com todos os El Niños "gigantes" desde 1972, com a temperatura que cada um atingiu.  
> **Técnico:** Anomalias de pico semanal (Niño 3.4) compiladas de boletins NOAA/CPC via INPE/INMET/MetSul.

**Fonte:** levantamento da MetSul Meteorologia com base em dados semanais da NOAA/CPC (ago/2026); confirmado por INPE/CPTEC e O Presente (24/mai/2026).


```python
eventos = pd.DataFrame([
    {'evento': '1972–73', 'ano': 1972, 'pico_c': 2.0,  'status': 'Muito forte'},
    {'evento': '1982–83', 'ano': 1982, 'pico_c': 2.6,  'status': 'Super El Niño'},
    {'evento': '1997–98', 'ano': 1997, 'pico_c': 2.3,  'status': 'Super El Niño'},
    {'evento': '2015–16', 'ano': 2015, 'pico_c': 3.0,  'status': 'Super El Niño (recorde)'},
    {'evento': '2023–24', 'ano': 2023, 'pico_c': 2.1,  'status': 'Muito forte'},
    {'evento': '2026–27', 'ano': 2026, 'pico_c': 2.6,  'status': 'Super El Niño (em curso)'},
])
eventos['intervalo_anos'] = eventos['ano'].diff()
display(eventos)
```

```python
fig, ax = plt.subplots(figsize=(10, 5.5))
cores = ['#FFB74D' if 'em curso' in s else ('#EF5350' if 'recorde' in s else '#FF7043') for s in eventos['status']]
bars = ax.bar(eventos['evento'], eventos['pico_c'], color=cores, edgecolor='white', linewidth=1.2)
ax.axhline(2.0, color='#455A64', linestyle='--', linewidth=1.3, label='Limiar "Super El Niño" (ONI ≥ 2,0°C)')
for b, v in zip(bars, eventos['pico_c']):
    ax.text(b.get_x()+b.get_width()/2, v+0.05, f"+{v:.1f}°C", ha='center', fontweight='bold', fontsize=10)
ax.set_ylabel("Anomalia de pico — Região Niño 3.4 (°C)")
ax.set_title("Episódios de El Niño Muito Forte / Super El Niño desde 1972\n(dados NOAA/CPC, compilação MetSul/INPE)")
ax.set_ylim(0, 3.4)
ax.legend(loc='upper left', fontsize=9)
plt.tight_layout()
plt.savefig('fig01_historico_super_elnino.png', bbox_inches='tight')
plt.show()
```

📊 **O que este gráfico mostra:** barras com a anomalia máxima de temperatura (°C) atingida pela região Niño 3.4 em cada episódio de El Niño muito forte desde 1972. A linha tracejada marca o limiar oficial de "Super El Niño" (+2,0°C).

**Leitura:** o evento de 2026–27 já empata com o segundo maior pico da série (1982-83) e ainda está em desenvolvimento — modelos da NOAA apontam possibilidade de superar o recorde de 2015-16 (+3,0°C) até o fim do ano.

## ⏱️ Seção 3 — Intervalo entre Super El Niños: um padrão de aceleração
> **Leigo:** Os El Niños gigantes estão ficando mais frequentes.  
> **Técnico:** Diferença em anos entre eventos consecutivos, evidenciando redução do intervalo médio.


```python
fig, ax = plt.subplots(figsize=(9, 5))
interv = eventos.dropna(subset=['intervalo_anos'])
periods = [f"{int(a)}→{int(b)}" for a,b in zip(eventos['ano'][:-1], eventos['ano'][1:])]
ax.plot(periods, interv['intervalo_anos'], 'o-', color='#D32F2F', linewidth=2.5, markersize=9)
for x, y in zip(periods, interv['intervalo_anos']):
    ax.text(x, y+0.5, f"{int(y)} anos", ha='center', fontweight='bold', fontsize=10)
ax.set_ylabel("Intervalo entre eventos (anos)")
ax.set_title("Encurtamento do intervalo entre El Niños muito fortes\n(1972–2026, fonte: NOAA/CPC)")
ax.set_ylim(0, 20)
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig('fig02_intervalo_eventos.png', bbox_inches='tight')
plt.show()
```

📊 **O que este gráfico mostra:** o número de anos entre um Super El Niño e o próximo, ao longo do tempo — quanto menor a linha, mais frequentes os eventos estão ficando.

**Leitura:** o intervalo caiu de 18 anos (1997→2015) para 8 anos (2015→2023) e, se 2026-27 se confirmar como muito forte, para apenas 3 anos desde 2023-24 — o menor intervalo da série desde 1950.

> **Leigo:** Como o fenômeno foi crescendo mês a mês este ano, segundo os boletins oficiais.  
> **Técnico:** Evolução da anomalia de TSM Niño 3.4 e da probabilidade declarada pela NOAA/CPC a cada boletim.

## 🎲 Seção 6 — Probabilidade de um Super El Niño nesta Década (2020–2029)
> **Leigo:** Qual a chance de um evento desse porte acontecer numa década qualquer — e o que já aconteceu de fato nesta década.  
> **Técnico:** Modelo de Poisson com taxa histórica de base (1950–2026) comparado à contagem empírica observada em 2020–2029.


📊 **O que este gráfico mostra:** barras com o número de episódios muito fortes/Super El Niño registrados em cada década desde 1950. A linha tracejada marca a taxa esperada pelo modelo histórico de Poisson.


```python
n_eventos = 5   # 1972, 1982, 1997, 2015, 2023 (excluindo 2026, ainda em curso)
anos_periodo = 2026 - 1950
taxa_anual = n_eventos / anos_periodo
prob_por_decada_poisson = 1 - np.exp(-taxa_anual*10)

eventos_por_decada = {'1950s':0,'1960s':0,'1970s':1,'1980s':1,'1990s':1,'2000s':0,'2010s':1,'2020s':2}
xs = list(eventos_por_decada.keys())
ys = list(eventos_por_decada.values())

fig, ax = plt.subplots(figsize=(9.5, 5.5))
cores5 = ['#EF5350' if v>=2 else ('#FFA726' if v==1 else '#90A4AE') for v in ys]
bars5 = ax.bar(xs, ys, color=cores5, edgecolor='white')
for b,v in zip(bars5, ys):
    ax.text(b.get_x()+b.get_width()/2, v+0.05, str(v), ha='center', fontweight='bold')
ax.axhline(taxa_anual*10, color='#455A64', linestyle='--', linewidth=1.3,
           label=f'Taxa histórica esperada (~{taxa_anual*10:.2f} evento/década)')
ax.set_ylabel("Nº de episódios muito fortes / Super El Niño")
ax.set_title(f"Episódios muito fortes por década (1950–2026)\nP(≥1 em 10 anos) via Poisson ≈ {prob_por_decada_poisson:.0%} | Década de 2020: já são 2 eventos")
ax.legend(fontsize=8.5)
plt.tight_layout()
plt.savefig('fig05_probabilidade_decada.png', bbox_inches='tight')
plt.show()

print(f"\n📐 Taxa histórica (1950–2026): {n_eventos} eventos em {anos_periodo} anos → {taxa_anual:.3f}/ano")
print(f"   P(≥1 Super El Niño em 10 anos) ≈ {prob_por_decada_poisson:.1%}")
print(f"   Observado em 2020–2029: 2 eventos (2023-24 e 2026-27) — acima da taxa histórica esperada")
```

## 🗺️ Seção 7 — Impactos no Brasil: o Mosaico Completo (não só a Amazônia)
> **Leigo:** El Niño não é só "seca na Amazônia" — cada região do Brasil sente o fenômeno de um jeito diferente, e alguns dos maiores riscos (energia, calor, enchente) estão fora da Amazônia.  
> **Técnico:** Matriz de intensidade por região × indicador, construída a partir de boletins INMET (jun-ago/2026), CNN Brasil (previsão Climatempo/Nottus, jun/2026), Greenpeace Brasil e análise setorial do Neo Mondo (jun/2026).


```python
regioes = ['Norte', 'Nordeste', 'Centro-Oeste', 'Sudeste', 'Sul']
indicadores = ['Anomalia de chuva\n(– seca / + excesso)', 'Risco de onda\nde calor', 'Risco agrícola\n(sequeiro/fungos)', 'Risco hídrico/energia\n(hidrelétricas)', 'Risco de\nenchente/deslizamento']

matriz = np.array([
    [-2, 1, 2, 1, 0],   # Norte
    [-2, 1, 2, 0, 0],   # Nordeste
    [-1, 2, 1, 2, 0],   # Centro-Oeste
    [-1, 2, 1, 2, 0],   # Sudeste
    [ 2, 0, 1, 0, 2],   # Sul
])

fig, ax = plt.subplots(figsize=(10.5, 5.5))
sns.heatmap(matriz, annot=True, fmt='d', cmap='RdBu_r', center=0, vmin=-2, vmax=2,
            xticklabels=indicadores, yticklabels=regioes, cbar_kws={'label': 'Intensidade do efeito'},
            linewidths=1, linecolor='white', ax=ax, annot_kws={'size': 12, 'weight':'bold'})
ax.set_title("Impactos esperados do El Niño 2026 por região do Brasil\n(mosaico completo — 2º semestre de 2026)")
plt.xticks(rotation=15, ha='right')
plt.tight_layout()
plt.savefig('fig06_matriz_regioes_brasil.png', bbox_inches='tight')
plt.show()
```

📊 **O que este gráfico mostra:** mapa de calor com 5 regiões do Brasil (linhas) × 5 indicadores de impacto (colunas). Azul-escuro = déficit/risco negativo forte; vermelho-escuro = excesso/risco positivo forte.

**Leitura região por região:**

- **Norte e Nordeste:** o padrão "clássico" — déficit de chuva, veranicos, risco elevado para lavouras de sequeiro (soja, milho) e maior frequência de ondas de calor a partir de setembro.
- **Centro-Oeste e Sudeste:** aqui está o risco menos falado e um dos mais sérios — **ondas de calor combinadas com baixa umidade**, e um **risco sistêmico para o setor elétrico**: essas duas regiões concentram cerca de 70% da capacidade hidrelétrica instalada do país, então seca + calor pode reduzir o nível dos reservatórios, forçar o acionamento de termelétricas mais caras e pressionar a tarifa de energia e o preço de alimentos. Bloqueios atmosféricos também podem represar o calor por períodos longos, com risco à saúde pública (a OMS aponta ondas de calor como uma das principais causas de morte relacionada ao clima).
- **Sul (RS, SC, PR):** o efeito se inverte — chuva acima da média, maior risco de **enchentes e deslizamentos**, encharcamento do solo e doenças fúngicas em cereais de inverno. O Rio Grande do Sul, em particular, concentra o maior risco de eventos extremos de precipitação da região.

## 🌎 Seção 8 — Impactos nos Países Vizinhos
> **Leigo:** O mesmo El Niño que seca o Nordeste também está mudando o clima na Argentina, no Peru, na Colômbia e em quase toda a América do Sul — cada país de um jeito.  
> **Técnico:** Padrão de anomalia (seca/calor vs. chuva/enchente) por país ou sub-região, compilado de SMN, SENAMHI/ENFEN, DMC, CNN Brasil e do mapa de teleconexões ENSO do IRI/Columbia University.

## 🌍 Seção 9 — Sismicidade Recente na América do Sul (dados reais, USGS)
> **Leigo:** Terremotos fortes que atingiram a América do Sul em 2026, com magnitude e número de vítimas.  
> **Técnico:** Importação direta dos dados via API pública do USGS (United States Geological Survey), com fallback para tabela verificada manualmente caso a rede não esteja disponível no ambiente de execução.

**Nota sobre acesso a dados:** a célula abaixo tenta puxar os dados **ao vivo** direto da API do USGS via `pandas.read_csv()`. Isso funciona normalmente no Google Colab (que tem internet liberada). Caso a rede esteja bloqueada no seu ambiente, o código usa automaticamente uma tabela de fallback com os eventos confirmados mais relevantes de 2026, verificados via USGS/Wikipedia/imprensa (Reuters, CNN, EFE) em agosto/2026.


```python
import pandas as pd

# ── 9.1 Tentativa de importação AO VIVO via API pública do USGS ──
# Documentação: https://earthquake.usgs.gov/fdsnws/event/1/
usgs_url = (
    "https://earthquake.usgs.gov/fdsnws/event/1/query?"
    "format=csv&starttime=2026-01-01&endtime=2026-08-27"
    "&minmagnitude=6.0"
    "&minlatitude=-56&maxlatitude=13&minlongitude=-82&maxlongitude=-34"  # bounding box América do Sul
)

try:
    terremotos = pd.read_csv(usgs_url)
    terremotos = terremotos[['time', 'latitude', 'longitude', 'depth', 'mag', 'place']]
    terremotos.columns = ['data_hora_utc', 'lat', 'lon', 'profundidade_km', 'magnitude', 'local']
    fonte = 'USGS (ao vivo via pandas.read_csv)'
    print(f"✅ {len(terremotos)} eventos M≥6.0 importados ao vivo da API do USGS")
except Exception as e:
    print(f"⚠️ Não foi possível acessar a API do USGS neste ambiente ({type(e).__name__}).\n"
          f"   Usando tabela de fallback com eventos verificados manualmente (USGS/Wikipedia/Reuters/CNN, ago/2026).\n")
    terremotos = pd.DataFrame([
        {'data_hora_utc': '2026-06-24 22:04', 'lat': 10.371, 'lon': -68.556, 'profundidade_km': 21.9, 'magnitude': 7.2, 'local': 'Yaracuy, Venezuela (precursor)'},
        {'data_hora_utc': '2026-06-24 22:05', 'lat': 10.622, 'lon': -67.194, 'profundidade_km': 10.0, 'magnitude': 7.5, 'local': 'La Guaira, Venezuela (principal)'},
        {'data_hora_utc': '2026-08-10 10:59', 'lat': 4.9,   'lon': -76.3,   'profundidade_km': 110.3,'magnitude': 7.4, 'local': 'Chocó, Colômbia'},
        {'data_hora_utc': '2026-08-20 10:28', 'lat': -14.0, 'lon': -73.4,   'profundidade_km': 67.0, 'magnitude': 6.7, 'local': 'Apurímac, Peru'},
    ])
    fonte = 'Fallback verificado manualmente (USGS/Wikipedia/Reuters/CNN, ago/2026)'

print(f"Fonte dos dados: {fonte}")
display(terremotos.sort_values('magnitude', ascending=False))
```

## 💥 Seção 10 — Efeito Combinado: Terremotos + El Niño 2026 (vidas, mercado, saúde, habitação)
> **Leigo:** os terremotos e o El Niño de 2026 têm causas físicas totalmente diferentes e um não provoca o outro — mas quando os dois atingem os mesmos países no mesmo ano, os efeitos se somam. Aqui olhamos o que isso significa na prática: vidas, dinheiro, saúde e moradia.  
> **Técnico:** análise de impacto composto (não causal) usando dados verificados de PAHO/OMS, CNN Brasil, Revista Fórum e Verisk para os terremotos, cruzados com o padrão de seca/calor do El Niño já identificado na Seção 8 para a mesma região (norte da América do Sul).

**Nota sobre causalidade (rápida, e depois seguimos para o que importa):** não há consenso científico ligando El Niño a terremotos — os mecanismos são independentes (tectônica de placas vs. oceano-atmosfera). O que segue **não é sobre causa**, é sobre **dois choques reais acontecendo nos mesmos países, ao mesmo tempo, sobre a mesma população**.


```python
hospitais = pd.DataFrame([
    {'hospital': 'Dr. Rafael Medina\nJiménez (La Guaira)', 'antes': 108, 'depois': 35},
])
fig, ax = plt.subplots(figsize=(7, 5.5))
x = np.arange(len(hospitais))
w = 0.35
ax.bar(x - w/2, hospitais['antes'], w, label='Leitos antes do terremoto', color='#1565C0')
ax.bar(x + w/2, hospitais['depois'], w, label='Leitos após o terremoto', color='#B71C1C')
ax.set_xticks(x)
ax.set_xticklabels(hospitais['hospital'])
ax.set_ylabel("Leitos disponíveis")
ax.set_title("Perda de capacidade hospitalar após o terremoto\nVenezuela, jun/2026 (dado real: PAHO/CNN Brasil)")
for i, (a, d) in enumerate(zip(hospitais['antes'], hospitais['depois'])):
    ax.text(i-w/2, a+2, str(a), ha='center', fontweight='bold')
    ax.text(i+w/2, d+2, str(d), ha='center', fontweight='bold')
    ax.text(i, max(a,d)+12, f"–{(1-d/a):.0%}", ha='center', color='#B71C1C', fontweight='bold', fontsize=12)
ax.legend()
ax.set_ylim(0, 140)
plt.tight_layout()
plt.savefig('fig09_hospital_capacidade.png', bbox_inches='tight')
plt.show()
```

📊 **O que este gráfico mostra:** capacidade de internação de um hospital-referência de La Guaira antes e depois do terremoto de junho/2026 — o hospital de maior prioridade segundo OMS/OPAS na região perdeu mais de dois terços de seus leitos.

**Por que isso importa junto com o El Niño:** o 2º semestre de 2026 traz calor recorde para exatamente essa região (Seção 8) — e ondas de calor aumentam a demanda por atendimento de emergência justamente quando a rede hospitalar de La Guaira/Caracas está operando com uma fração da capacidade normal, sem ventiladores mecânicos funcionando e com banco de sangue crítico.

3. **Quatro institutos sul-americanos convergem no mesmo diagnóstico**: Brasil (92%), Argentina (90%), Chile (63%) e Peru (50%+) todos apontam intensificação do fenômeno até o fim de 2026.
4. **Probabilidade de um Super El Niño nesta década:** o modelo histórico de longo prazo (1950–2026) sugere ~48% de chance por década — mas a década de 2020 já teve **dois** eventos muito fortes, acima da expectativa histórica.
5. **Impacto no Brasil vai muito além da Amazônia (Seção 7):** Norte/Nordeste com seca clássica e veranicos; Centro-Oeste/Sudeste com ondas de calor + risco sistêmico ao setor elétrico (essas duas regiões concentram ~70% da hidrelétrica nacional); Sul com chuva acima da média e risco de enchente/deslizamento — três riscos bem diferentes acontecendo ao mesmo tempo, não só "seca na Amazônia".
6. **Não é só o Brasil (Seção 8):** a Bacia do Prata (Argentina norte/nordeste, Paraguai, Uruguai) segue o mesmo padrão do Sul brasileiro — chuva em excesso e risco de cheia de rios compartilhados (180 mil desalojados simultaneamente em 2015-16). Peru e Equador vivem o epicentro físico do fenômeno, com calor extremo e huaicos. Colômbia e Venezuela, na ponta oposta do continente, enfrentam seca e calor recorde.

7. **Sismicidade 2026 (Seção 9):** três terremotos fortes atingiram a América do Sul em 2026 — Venezuela (Mw 7,2+7,5, jun/2026, balanço mais recente de 6.509 mortos e 226.696 atendimentos hospitalares), Colômbia (Mw 7,4, ago/2026, 179-331 mortos) e Peru (Mw 6,7, ago/2026, sem vítimas fatais). Isso está dentro da faixa normal global (11 sismos M≥7 até agosto, contra 11-20/ano desde 2020) — 2026 não é um ano sismicamente atípico.
8. **Efeito combinado, não causal (Seção 10):** não há consenso científico ligando El Niño a terremotos — mecanismos independentes (tectônica vs. oceano-atmosfera). Mas o impacto real e mensurável é a **soma de vulnerabilidades**: a Venezuela perdeu 68% dos leitos de um hospital-chave de La Guaira, tem 600 MW de capacidade elétrica fora do ar e 24.000+ pessoas em acampamentos — tudo isso justamente quando entra na fase de calor recorde e possível seca hidrológica do El Niño no mesmo semestre, o que agrava vidas, mercado, saúde e habitação ao mesmo tempo.

**Limitações e transparência:**
- Os dados de picos semanais usam a série da NOAA (não o ONI trimestral oficial) — métrica amplamente usada pela imprensa meteorológica brasileira, mas que não deve ser confundida com a classificação ONI formal.
- 5 dos 11 institutos sul-americanos solicitados (DMH-Paraguai, IDEAM-Colômbia, INAMHI-Equador, INUMET-Uruguai, SENAMHI-Bolívia) não tinham, nas fontes consultadas, boletins ENSO quantitativos específicos de 2026 — dado esse limite, não foram incluídos com número próprio no painel.
